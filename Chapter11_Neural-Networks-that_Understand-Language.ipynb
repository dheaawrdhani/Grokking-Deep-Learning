## The big picture

Chapter 11 introduces **natural language processing (NLP)** through supervised sentiment analysis and word embeddings. The key idea is that a neural network can learn useful representations of words from how those words correlate with a task's labels.

## 1. What does it mean to understand language?

Language understanding is approached as a prediction problem: provide text as input and train a model to predict an output, such as whether a movie review is positive or negative. The chapter focuses on patterns in language rather than hand-written rules.

## 2. Supervised NLP and the IMDB dataset

The IMDB movie-review dataset pairs reviews with sentiment labels. A model learns from examples whose correct labels are known, then uses the learned relationships to predict sentiment for unseen reviews. Text must be converted into numerical input before a neural network can process it.

## 3. Bag-of-words and word correlation

A **bag-of-words** representation records which words occur in a review, often using counts or presence indicators. It is simple and useful, but it largely ignores word order and context: “not good” and “good” can look too similar if only word presence is considered.

```python
# Tiny bag-of-words example (illustrative)
vocabulary = ['great', 'boring', 'not']
review = 'not great'
vector = [int(word in review.split()) for word in vocabulary]
print('Vocabulary:', vocabulary)
print('Review vector:', vector)
```

## 4. Embedding layers

An **embedding** maps each word to a dense vector of learned numbers. Instead of representing a word only as a large sparse indicator, the network learns a compact representation that can capture task-relevant relationships. Similarity between embeddings can be inspected with vector comparisons.

## 5. What do neurons and embeddings mean?

An embedding does not have a fixed human-defined meaning for each coordinate. Its values are learned because they help reduce the model's loss on the task. A dimension may reflect a useful pattern, but the meaning of the representation comes from how the full model uses it.

## 6. Word analogies

The chapter explores relationships such as **king − man + woman ≈ queen**. Such vector arithmetic can emerge when embeddings encode regularities in word relationships. It is an approximate pattern, not a guarantee that every analogy works or that the model understands language exactly as a person does.

## Summary

- NLP applies machine learning to language tasks.
- Supervised sentiment analysis learns from labeled reviews.
- Bag-of-words is simple but loses word order and context.
- Embedding layers learn dense, task-useful word representations.
- Embedding dimensions gain their usefulness through training and loss minimization.
- Vector relationships can encode analogies and similarities.
