## The big picture

Chapter 12 addresses text whose length varies from example to example. It develops sentence representations and recurrent neural networks (RNNs), which process a sequence step by step while carrying information forward.

## 1. The challenge of arbitrary-length text

A fixed-size input layer expects the same number of input values each time, but sentences contain different numbers of words. A model therefore needs a way to represent variable-length sequences without discarding all information about how words relate to one another.

## 2. Averaged word vectors and bag-of-words limitations

Averaging word embeddings creates a fixed-size sentence vector regardless of sentence length. This is simple and can work surprisingly well, but it loses word order: sentences containing the same words can receive the same or very similar representation even when their meanings differ.

```python
# Illustrative averaging of word vectors
import numpy as np
embeddings = {
    'deep': np.array([1.0, 0.0]),
    'learning': np.array([0.0, 2.0]),
    'works': np.array([1.0, 1.0])
}
sentence = ['deep', 'learning', 'works']
sentence_vector = np.mean([embeddings[w] for w in sentence], axis=0)
print(sentence_vector)
```

## 3. Recurrent processing and transition matrices

An RNN processes one word at a time. At each step, it combines the current word representation with a hidden state carrying information from earlier words. A learned transition transformation updates that state, allowing the representation to depend on sequence order.

## 4. Sentence vectors learned from a task

Rather than simply adding embeddings, the model learns transformations that make the resulting sentence representation useful for prediction. Training adjusts the embeddings and transition weights so the representation captures patterns associated with the target task.

## 5. Forward propagation and backpropagation for variable lengths

During forward propagation, the model repeatedly updates its hidden state for each token. During backpropagation, the prediction error is traced through the sequence to adjust the shared weights. Because the same transformations are reused at each step, learning can improve how the network handles sequences of different lengths.

## Summary

- Text sequences have variable lengths, which complicates fixed-size inputs.
- Averaged embeddings produce fixed-size vectors but lose word order.
- RNNs process tokens sequentially and carry a hidden state forward.
- Transition matrices learn how information from earlier tokens affects later states.
- Backpropagation through the sequence trains shared weights for sequence modeling.
