## The big picture

Chapter 13 builds a small deep learning framework to clarify what libraries such as PyTorch do internally. The core ideas are tensors, automatic gradient computation (autograd), layers, loss functions, and optimization.

## 1. Tensors and frameworks

A **tensor** is a numerical container that generalizes scalars, vectors, and matrices to multiple dimensions. A deep learning framework organizes tensor operations and model layers so that predictions, gradients, and parameter updates can be computed consistently.

```python
# Tensor shapes with NumPy (illustrative)
import numpy as np
scalar = np.array(3.0)
vector = np.array([1.0, 2.0, 3.0])
matrix = np.array([[1.0, 2.0], [3.0, 4.0]])
print('scalar shape:', scalar.shape)
print('vector shape:', vector.shape)
print('matrix shape:', matrix.shape)
```

## 2. Automatic gradient computation (autograd)

Training requires derivatives that show how a change in a parameter affects the loss. Autograd records the operations used to calculate a result and applies the chain rule in reverse to calculate gradients. When a tensor is used multiple times, its gradient contributions from all paths must be combined.

## 3. Backpropagation through operations

Each supported operation needs a corresponding backward rule. For example, addition passes gradient contributions to both operands; negation changes the gradient's sign. More complex operations require their own gradient rules. This is how a computation graph becomes differentiable end to end.

## 4. Layers, loss, and optimization

Layers package parameters and operations. The chapter adds building blocks such as nonlinearities, embeddings, cross-entropy loss, and a recurrent layer. An optimizer uses calculated gradients to update model parameters—for example, stochastic gradient descent moves weights in the direction that reduces loss.

## 5. Why build a framework yourself?

Implementing a small framework connects the concepts from earlier chapters to the abstractions used in practical libraries. It also reveals that a framework is not magic: it combines tensor storage, forward operations, gradient rules, layers, loss calculations, and parameter updates.

## Summary

- Tensors hold numerical data of different dimensions.
- Autograd computes gradients by tracking operations and applying the chain rule.
- Each operation needs a correct backward rule; reused tensors accumulate gradient contributions.
- Layers combine operations and parameters into reusable components.
- Loss functions measure prediction error, and optimizers update weights using gradients.
- Understanding these pieces makes modern frameworks easier to use effectively.
