## Long short-term memory (LSTM)

**Companion notebook for *Grokking Deep Learning* by Andrew W. Trask**

This notebook summarizes the chapter's main ideas in plain language, with small illustrative examples where useful.

## The big picture

Chapter 14 extends recurrent networks to character-level language modeling. The model learns patterns in a text corpus and predicts the next character, enabling it to generate text one character at a time.

## 1. Character language modeling

The text is represented as a sequence of characters. Given previous characters, the network predicts a distribution over the next character. Repeatedly choosing a predicted character and feeding it back as the next input produces a longer sample. The generated text can resemble the style of the training corpus without reproducing coherent human-level writing.

## 2. Truncated backpropagation through time

Backpropagating through a very long sequence is computationally expensive and difficult to manage. **Truncated backpropagation** divides a sequence into shorter chunks and updates the model using each chunk. The hidden state can carry context between chunks, while gradients are only propagated through a limited span.

## 3. Vanishing and exploding gradients

In recurrent networks, gradients are repeatedly propagated across time steps. Repeated multiplication can make them shrink toward zero (**vanishing gradients**) or grow excessively (**exploding gradients**). Vanishing gradients make long-range dependencies hard to learn; exploding gradients can destabilize training.

## 4. LSTM cells and gates

**Long short-term memory (LSTM)** introduces a cell state and gates that regulate what information is kept, added, or exposed. The gates provide a controlled path for information to persist across many time steps, helping mitigate the vanishing-gradient problem and supporting longer-term dependencies.

## 5. Training and tuning the character model

The model is trained to predict the next character from context. Training settings and sampling choices affect output quality. The chapter's Shakespeare-style examples demonstrate that a model can learn statistical patterns such as character sequences, punctuation, and dialogue formatting, even when the output is not always grammatical or meaningful.

## Summary

- Character-level language models predict the next character in a sequence.
- Truncated backpropagation makes training on long text more manageable.
- RNNs can suffer from vanishing and exploding gradients.
- LSTM gates regulate information flow through a cell state and help preserve longer-term context.
- Generated text reflects learned statistical patterns, but it may still contain errors or nonsense.
