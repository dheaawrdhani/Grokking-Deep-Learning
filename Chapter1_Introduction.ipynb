## The big picture

Chapter 1 is a pep talk with a purpose. Before any math or code, the author explains *why* deep learning deserves effort and *how* the book intends to teach it. Three claims carry the argument:

1. Deep learning is a practical way to automate intelligence a little at a time.

---
## Why learn deep learning?

### 1. A tool for automating intelligence, step by step
People have always built tools to understand and control their surroundings. Deep learning is the latest tool in that line, and it is unusual because it is *mental* rather than mechanical: it automates pattern recognition and decision making.
The remarkable part is how uniform it is. Computer vision, speech recognition, and machine translation all got their record-breaking results from the same family of brain-inspired models, **neural networks**.

### 2. It can automate skilled labor
The author is careful to separate hype from a real concern. Even if research stopped today, current systems could already provide cheap substitutes for call-center operators, drivers, and entry-level analysts. The suggested response is to move toward the field that is growing, i.e. to learn the technology.

### 3. It is creative
Teaching a machine to paint like Monet, or to read sentiment in text, forces you to ask what thinking, reasoning, and creating actually involve. That curiosity is a big part of what keeps people in the field.

```python
import sys
import numpy as np
import matplotlib
print("Python     :", sys.version.split()[0])
print("NumPy      :", np.__version__)
print("Matplotlib :", matplotlib.__version__)
```

```python
# Quick sanity check of the two operations the book relies on most:
# elementwise multiplication and the dot product.
a = np.array([1.0, 2.0, 3.0])
b = np.array([0.5, 0.5, 0.5])

print("a * b   =", a * b)        # elementwise: [0.5 1.  1.5]
print("a . b   =", a.dot(b))     # multiply then sum: 3.0
```

If both cells ran without errors, you are ready. The two operations above (`*` and `.dot`) are literally the engine of a neural network's prediction step in Chapter 3.
